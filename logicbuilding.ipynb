# Phase 1 : While Loop

1. Print all numbers from 1 to 10 using a loop.

```python
i = 1
while i<=10:
    print(i,end=" ")
    i+=1
```

2. Print numbers from 10 down to 1 in reverse order.

```python
i=10
while i>=1:
    print(i,end=" ")
    i-=1
```

3. Print all even numbers between 1 and 100.

```python
print("Even Numbers from 1 to 100 are:\n")
i = 1
while i<=100:
    if i%2==0:
        print(i,end=" ")
    i+=1
```

4. Print all odd numbers between 1 and 100.

```python
print("Odd Numbers from 1 to 100:\n")
i = 1
while i<=100:
    if i % 2==1:
        print(i,end=" ")
    i +=1
```

5. Print the multiplication table of 5

```python
i = 5
j=1
while j<=10:
    print("5","*",j,"=",i*j)
    j +=1
```

6. Print the multiplication table of a given number from n × 1 to n × 10.

```python
i = int(input("Enter the number which table you want to print"))
j=1
while j<=10:
    print(i,"*",j,"=",i*j)
    j +=1
```

1. Print all numbers from 1 to 10 using a loop.

```python
i = 1
while i<=10:
    print(i,end=" ")
    i+=1
```

11. Find and print the product of all digits of a given number.

```python
n = int(input("Enter your number: "))
product = 1
n = abs(n)

if n == 0:
    product = 0
else:
    while n > 0:
        rem = n % 10
        product = product * rem
        n = n // 10
print("The product is ",product)
```

14. Check whether the given number is a palindrome.

```python
n = int(input("Enter your number:"))
original = n
rev = 0
while n>0:
    rem = n%10
    rev = rev *10 + rem
    n =n//10
if original == rev :
    print("The number is palindrome")
else:
    print("The number is not palindrome")
```

16. Check whether the given number is an Armstrong number.

```python
n = int(input("Enter your number:"))
original = n
number_len = str(n)
power = len(number_len)
sum = 0
while n>0:
    rem = n%10
    sum = sum + rem ** power
    n =n//10
if original == sum :
    print("The number is Armstrong Number")
else:
    print("The number is not Armstrong Number")
```

17. Check whether the given number is a Perfect number.

```python
n = int(input("Enter your number"))
sum = 0
original = n
i =1
while i<n-1:
    if (n % i ==0):
        sum = sum + i
    i +=1
if sum == original :
    print("The given number is a perfect number ")
else:
    print("The given number is not a perfect number")
```
